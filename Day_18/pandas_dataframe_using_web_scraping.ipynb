```python
final=pd.DataFrame()

for page in range(1,51):
    url = f'https://books.toscrape.com/catalogue/page-{page}.html'
    webpage=requests.get(url).text
    soup=BeautifulSoup(webpage,'lxml')
    books=soup.find_all('article',class_='product_pod')

    name=[]
    price=[]
    availability=[]
    rating=[]

    for i in books:
        name.append(i.find('h3').find('a').get('title'))
        price.append(i.find('p',class_='price_color').text.strip())
        availability.append(i.find('p',class_='instock availability').get_text(" ",strip=True))
        rating.append(i.find('p',class_='star-rating').get('class')[1])

    temp=pd.DataFrame({
            'name':name,
            'price':price,
            'availability':availability,
            'rating':rating
        })
    final=pd.concat([final,temp],ignore_index=True)
    
```
