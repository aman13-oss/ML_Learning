```python
for i in range(1, 4):
    url = f"https://example-data.com/api/v1/users?limit=100&page={i}"
    response=requests.get(url)
    team_df=pd.DataFrame(response.json()['data'])
    df=pd.concat([df,team_df],ignore_index=True)
    
```
